```python
# logical operators
# and , or , not

# and table

# T and T = T
# T and F = F
# F and T = F
# F and F = F


# or table
# T or T = T
# T or F = T
# F or T = T
# F or F = F


# not

# T and T = not(T) = F
# T and F = not(F) = T
# F and T = not(F) = T
# F and F = not(F) = T


a=3
b=2

print(a!=b and a > b) # T and T = T # no short circuit
print(a!=b and a==b) # T and F = F  # no short circuit
print(a==b and a!=b) # F and T = F  # short circuit
print(a==b and a<b) # F and F = F # short circuit

print("=================================")

print(a!=b or a > b) # T or T = T # short circuit
print(a!=b or a==b) # T or F = T  # short circuit
print(a==b or a!=b) # F or T = T  # no short circuit
print(a==b or a<b) # F or F = F # no short circuit

print("=================================")

print(not(a!=b and a > b)) # T and T = not(T) = F # no short circuit
print(not(a!=b and a==b)) # T and F = not(F) = T  # no short circuit
print(not(a==b and a!=b)) # F and T = not(F) = T  # short circuit
print(not(a==b and a<b)) # F and F = not(F) = T # short circuit
```

```python
# functions are used tpo reusable peice of code


# syntax
# def funtioname(parameters(optional)):
      # statements
      # return (optinal)
      
def calculateexp(exp):  # [12300,45000,16000,34000]
    total=0
    for item in exp:   # for item  in [12300,45000,16000,34000]
        total=total+item
    
    return total

respraveen=calculateexp(praveenexp)
print(respraveen)

ressrinath=calculateexp(srinthexp)
print(ressrinath)

```

```python
# types of argumnets

# 1. positional arguments
# 2. default argumemts
# 3. keyord arguemnts
# 4. variable length arguments
# 5. variable length keyword arguments

# positional argumnets

def add(a,b):
    return a+b

res=add(4,2)
res

# keyword argumnets

def add(a,b,c):
    return a+b+c

res=add(b=5,c=4,a=2)
res

def add(a,b,c):
    # print(f"a value is {a}")
    # print(f"b values is {b}")
    return a+b+c

res=add(2,c=4,b=5)
res

# default arguments

def add(a,b=0):   
    return a+b

res=add(3,8)
res


# variable length arguments

def add(*args): # parameters tuple
    print(type(args))
    print(args)  # (2,3,7,8)
    total=0
    for item in args:
        total=total+item # total += item
    return total
             
    

result=add(2,3,7,8,9,10,12,17,95,34,89) # variable length argumnets
result


# variable length keyword arguments

print("====================================")
def studentinfo(**kwargs): # dictionary
    print(type(kwargs))
    print(kwargs) # {"name":"aruna","id":1,"address":"hyd"}
    for k,v in kwargs.items(): # for k,v in "address":"hyd"
        print(k,v)

    

res=studentinfo(id=1,address="hyd",name="aruna",phnum=986666,age=21)
res

#
```

#### Exception Handling 

```python

#  compile time errors
#  runtime error
try:
    a=int(input("enter number"))
    b=int(input("enter number"))
    a/b
    with open("test5.txt") as f:
        print(f.read())
except TypeError as e:
    print("error occured ", e)
except ZeroDivisionError as e:
    print("error occured ", e)
except FileNotFoundError as e:
    print("error occured ", e)
```

```python
try:
    a=int(input("enter number"))
    a/0
    with open("test5.txt") as f:
        a=f.read()
        # frontend logic
except Exception as e:
    print("error occured",e)
finally:
    # resource cleanup
    a=None
    print("execution completed")   
```
