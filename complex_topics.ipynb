```python
# comprehensions
# list , set , dictionary , tuple (default it is not there)

nums=[2,5,4,8,9]

# normal programmer

squares=[]
for item in nums:
    squares.append(item**2)
   
print(squares) 

# lazy programmers

#list comprehension syntax
# [expression forloop] 
result=[item**2 for item in nums]  # [5**2 for 5 in [2,5,4,8,9]]  # [4 25]
result


nums=[2,5,4,8,9]
# normal programmer

evensquares=[]
for item in nums:  #    4 in [2,5,4,8,9]
    if item%2==0:   # 4%2 == 0
        evensquares.append(item**2)
   
print(evensquares) 


# list comprehemsion with if conduition
# syntax
#[expression forloop if condition]
evenres=[item**2 for item in nums if item%2==0]
evenres



nums=[2,5,4,8,9]
# normal programmer

normalall=[]
for item in nums:  #    9 in [2,5,4,8,9]
    if item%2==0:   # 9%2 == 0
        normalall.append(item**2)   # [4 125 16 64]
    else:
        normalall.append(item**3)    # [4 125 16 64 729]
   
print(normalall) 

# lazy prgrammer

# syntax
#[expression ifcondition else expression forloop]


resall=[item**2 if item%2==0 else item**3 for item in nums]
resall
   
    
```

```python
# set comprehension

nums=[2,5,4,8,9]

# lazy programmers

#set comprehension syntax
# {expression forloop}
result={item**2 for item in nums}  
print(result)


# set comprehemsion with if conduition
# syntax
#{expression forloop if condition}
evenres={item**2 for item in nums if item%2==0}
print(evenres)


# lazy prgrammer

# syntax
#{expression ifcondition else expression forloop}


resall={item**2 if item%2==0 else item**3 for item in nums}
print(resall)
```

#### Decorator

```python
# decorator : it is used to change the behavior of teh function without modifying the existing function.


# def add(a,b):
#     # logging a and b
#     return a+b
def logindecorator(func):  # facebooklogin
    def wrapper():
        print("before login")
        func()  # facebooklogin()
        print("after login")
    return wrapper

@logindecorator
def facebooklogin():
    print("welcome to home page")
    
facebooklogin()

# befoire login
# welcome to home page
# after login
```
