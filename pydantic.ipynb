```python
from dataclasses import dataclass

@dataclass
class Employee:
    EmpName:str
    EmpId:int
    Salaary:float
    
empobj1=Employee("sai",1,20000)
print(empobj1.EmpName)
print(empobj1.EmpId)

print("===========================")

empobj2=Employee("naresh",2,2000.34)
print(empobj2.EmpName)
print(empobj2.EmpId)

print("===========================")
empobj3=Employee(12,"srikanth","abc")
print(empobj3.EmpName)
print(empobj3.EmpId)
```

```python
from pydantic import BaseModel
```

```python

class EmployeeNew(BaseModel):
    EmpName:str
    EmpId:int
    Salary:float
    
testobj=EmployeeNew(EmpName="test",EmpId=1,Salary=24000.45)
testobj

testobj1=EmployeeNew(EmpName="test1",EmpId="2",Salary=24000.45)  
testobj1
```

```python

from typing import List

class ClassRoom(BaseModel):
    roomnum:int
    students:List[str]
    capacity:int
    
clsobj=ClassRoom(roomnum="12",students=("sai","srikanth","naresh"),capacity=3) 
clsobj
```
